```python
import torch
import numpy as np
```

```python
def set_seed(seed):
    np.random.seed(seed)
    torch.manual_seed(seed)
```

```python
set_seed(42)
arr = np.random.randint(0, 5, size=6).astype('int32') # generates 6 random integers from 0 up to (but not included) 5

assert np.array_equal(arr, np.array([3,4,2,4,4,1])), "Oops"
print(arr)
```

```python
x = torch.from_numpy(arr)
```

```python
print('Old:', x.type())

x = x.type(torch.int64)

print('New:', x.type())
```

```python
x = x.reshape(3,2)
print (x)
```

```python
print(x[:,1:])
```

```python
sqrd = x.pow(2)
print(sqrd)
print(x)
```

```python
set_seed(42)
y = torch.randint(0,5,(2,3))

assert torch.equal(y, torch.tensor([[2, 2, 1],[4, 1, 0]])), "Oops"
print(y)
```

```python
prod = torch.mm(x,y)
prod
```
