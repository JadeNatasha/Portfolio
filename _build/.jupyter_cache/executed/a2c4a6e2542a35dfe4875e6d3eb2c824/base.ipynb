```python
import numpy as np

# Computing the weighted sums into the hidden layer (Z1, Z2)

x = np.array([1, 0, 1])          # inputs
W = np.array([
    [0.2, -0.3],    # weights from x1 -> [H1, H2]
    [0.4,  0.1],    # weights from x2 -> [H1, H2]
    [-0.5, 0.2]     # weights from x3 -> [H1, H2]
])
theta = np.array([-0.4, 0.2])    # biases for [H1, H2]

Z = x @ W + theta   # matrix multiply + add bias
```

```python
# Applying the activation function (ReLU) to get H1, H2
# ReLU (Rectified Linear Unit) rule: negative → 0, positive → keep it.

def relu(z):
    return np.maximum(0, z)

H = relu(Z)     # not relu(Z_hidden)
print("H1, H2 =", H)
```

```python
# Computing the weighted sum into the output layer (Z3)
# Now H1 and H2 are the inputs, and w21, w22 are the connecting weights.

w21, w22 = -0.3, -0.2
theta3 = 0.1

Z3 = (w21 * H[0]) + (w22 * H[1]) + theta3
print("Z3 =", Z3)
```

```python
# Applying activation to get the predicted output ŷ

y_hat = relu(Z3)
print("y_hat =", y_hat)
```

```python
# Computing the error
y_actual = 1

E = 0.5 * (y_actual - y_hat) ** 2
print("So the Error is E =", E)
```
